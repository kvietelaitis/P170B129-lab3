# 8. Training and validation

```python
import json
import matplotlib.pyplot as plt

def plot_run(run, ax):
    log = json.load(open(f"../outputs/{run}/log_history.json"))
    tr = [(r["step"], r["loss"]) for r in log if "loss" in r]
    va = [(r["step"], r["eval_loss"]) for r in log if "eval_loss" in r]
    ax.plot(*zip(*tr), label="train", alpha=0.7)
    ax.plot(*zip(*va), "o-", label="validation")
    best = min(va, key=lambda x: x[1])
    ax.axvline(best[0], ls="--", c="gray")
    ax.set_title(f"{run} (best val {best[1]:.3f} at step {best[0]})")
    ax.set_xlabel("step"); ax.set_ylabel("loss"); ax.legend()

runs = ["full", "lora"]
fig, axes = plt.subplots(1, len(runs), figsize=(6 * len(runs), 4))
for ax, r in zip(axes, runs):
    plot_run(r, ax)
plt.tight_layout(); plt.show()
```
