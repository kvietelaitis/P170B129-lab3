# 13. Overfitting

```python
import json, numpy as np, matplotlib.pyplot as plt

# 8 epochs

log = json.load(open("../outputs/long/log_history.json"))
tr = np.array([(r["step"], r["loss"]) for r in log if "loss" in r])
va = np.array([(r["step"], r["eval_loss"]) for r in log if "eval_loss" in r])
steps_per_epoch = max(r["step"] for r in log if "epoch" in r) / max(r["epoch"] for r in log if "epoch" in r)

best_i = va[:, 1].argmin()
best_step, best_loss = va[best_i]
print(f"Lowest validation loss {best_loss:.3f} at step {best_step:.0f} "
      f"(epoch {best_step / steps_per_epoch:.2f}); after that it rises.")
print(f"Final validation loss {va[-1, 1]:.3f} (+{va[-1, 1] - best_loss:.3f})")

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(tr[:, 0], tr[:, 1], label="train", alpha=0.7)
ax.plot(va[:, 0], va[:, 1], "o-", label="validation")
ax.axvline(best_step, ls="--", c="gray")
ax.axvspan(best_step, va[-1, 0], color="red", alpha=0.07)   # shade the overfitting region
for e in range(1, int(va[-1, 0] / steps_per_epoch) + 1):     # epoch boundaries
    ax.axvline(e * steps_per_epoch, c="black", alpha=0.1)
ax.set_xlabel("step"); ax.set_ylabel("loss"); ax.legend()
plt.show()
```
